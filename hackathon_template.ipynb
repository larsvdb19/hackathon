## 8. Code: data verkennen (EDA)

Zet het pad naar de aangeleverde data hieronder. Zonder data wordt er dummydata gemaakt zodat je de code kan testen.

```python
DATA_PATH = None  # bv. "data/verlof.csv"

if DATA_PATH:
    df = pd.read_csv(DATA_PATH)   # gebruik pd.read_excel voor .xlsx
else:
    # Dummydata om de template te testen (klantdata, past bij KBC-achtige cases)
    n = 1000
    df = pd.DataFrame({
        "leeftijd": np.random.randint(18, 80, n),
        "klant_sinds_jaren": np.random.gamma(2, 4, n).round(1),
        "segment": np.random.choice(["jong", "gezin", "senior", "kmo"], n),
        "transacties_per_maand": np.random.poisson(35, n),
        "app_logins_per_maand": np.random.poisson(12, n),
    })
    proba = 1 / (1 + np.exp(0.15 * df["app_logins_per_maand"] + 0.05 * df["klant_sinds_jaren"] - 2))
    df["opgezegd"] = (np.random.rand(n) < proba).astype(int)

print(df.shape)
df.head()
```

## 9. Code: baseline model

Begin altijd met een eenvoudige baseline. Pas `target` en `features` aan. Rapporteer ook onzekerheid, niet enkel een enkele score.

```python
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, roc_auc_score, average_precision_score

target = "opgezegd"                           # PAS AAN
features = [c for c in df.columns if c != target]  # PAS AAN (verwijder kolommen die het antwoord al verraden!)

X, y = df[features], df[target]
num = X.select_dtypes("number").columns.tolist()
cat = X.select_dtypes(exclude="number").columns.tolist()

prep = ColumnTransformer([
    ("num", StandardScaler(), num),
    ("cat", OneHotEncoder(handle_unknown="ignore"), cat),
])

X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.25, stratify=y, random_state=RANDOM_STATE)

modellen = {
    "Logistische regressie (baseline)": LogisticRegression(max_iter=1000),
    "Random forest": RandomForestClassifier(n_estimators=200, random_state=RANDOM_STATE),
}

for naam, m in modellen.items():
    pipe = Pipeline([("prep", prep), ("model", m)])
    cv = cross_val_score(pipe, X_tr, y_tr, cv=5, scoring="roc_auc")
    pipe.fit(X_tr, y_tr)
    auc = roc_auc_score(y_te, pipe.predict_proba(X_te)[:, 1])
    ap = average_precision_score(y_te, pipe.predict_proba(X_te)[:, 1])
    print(f"{naam}: CV AUC {cv.mean():.3f} ± {cv.std():.3f} | test AUC {auc:.3f} | AP {ap:.3f}")
```

**Let op:** bij bank- en verzekeringsdata is fraude/schade zeldzaam, dus kijk naar precision/recall (AP) en niet enkel AUC of accuracy. Controleer op data-lekkage (kolommen die het antwoord al bevatten), klasse-onbalans, en of het model eerlijk is voor subgroepen. Leg in de pitch uit *waarom* het model iets voorspelt (feature importance, voorbeelden).

## 10. Business case en impact

Jullie hebben geen economische opleiding, maar een eenvoudige berekening volstaat. Vul de aannames in (en verantwoord ze met een bron of redenering).

```python
# Voorbeeldberekening: tijdswinst door automatisering (PAS AAN; voor fraude/churn reken je met vermeden verlies i.p.v. uren)
aantal_gebruikers      = 200        # hoeveel mensen gebruiken dit?
taken_per_maand        = 20         # hoe vaak per gebruiker?
minuten_per_taak_nu    = 15
minuten_per_taak_nieuw = 3
uurloon_kost           = 40         # €/uur totale loonkost

bespaard_uren_per_jaar = aantal_gebruikers * taken_per_maand * 12 * (minuten_per_taak_nu - minuten_per_taak_nieuw) / 60
besparing_euro = bespaard_uren_per_jaar * uurloon_kost
ontwikkelkost  = 30000              # schatting eenmalig
jaarlijkse_kost = 5000              # onderhoud/hosting

netto_jaar1 = besparing_euro - ontwikkelkost - jaarlijkse_kost
print(f"Bespaarde uren/jaar: {bespaard_uren_per_jaar:,.0f}")
print(f"Besparing/jaar:      € {besparing_euro:,.0f}")
print(f"Netto jaar 1:        € {netto_jaar1:,.0f}")
print(f"Terugverdientijd:    {ontwikkelkost / max(besparing_euro - jaarlijkse_kost, 1) * 12:.1f} maanden")
```
